```python
import pandas as pd
```

```python
m200 = pd.read_csv('DATA/OUTPUT/MHW_summary_table_long_oisst.19812025.Portugal_200m_bathymetry.csv')
m400 = pd.read_csv('DATA/OUTPUT/MHW_summary_table_long_oisst.19812025.Portugal_400m_bathymetry.csv')
m600 = pd.read_csv('DATA/OUTPUT/MHW_summary_table_long_oisst.19812025.Portugal_600m_bathymetry.csv')
m800 = pd.read_csv('DATA/OUTPUT/MHW_summary_table_long_oisst.19812025.Portugal_800m_bathymetry.csv')
m1000 = pd.read_csv('DATA/OUTPUT/MHW_summary_table_long_oisst.19812025.Portugal_1000m_bathymetry.csv')
m200.head()
```

```python
m200['Bathymetry'] = 200
m400['Bathymetry'] = 400
m600['Bathymetry'] = 600
m800['Bathymetry'] = 800
m1000['Bathymetry'] = 1000
```

```python
concatenated = pd.concat([m200, m400, m600, m800, m1000])
concatenated.to_csv('DATA/OUTPUT/allBat_Concatenated_SummaryTable_long.csv', index= False)
concatenated.to_excel('DATA/OUTPUT/allBat_Concatenated_SummaryTable_long.xlsx', index= False)
```
